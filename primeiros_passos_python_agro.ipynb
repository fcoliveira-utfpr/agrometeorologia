## 2. Operadores numéricos

Os operadores numéricos são ferramentas essenciais para realizar operações matemáticas em números. Eles permitem que os programadores executem cálculos, façam comparações e manipulem números de forma eficiente. Aqui estão alguns dos operadores numéricos mais comuns em Python:

### Operadores Aritméticos:

*  `+` (adição): Realiza a soma de dois números.
*  `-` (subtração): Subtrai um número do outro.
* `*` (multiplicação): Multiplica dois números.
* `/` (divisão): Divide um número pelo outro.
* `%` (módulo/resto): Retorna o resto da divisão de um número pelo outro.
* `**` (potenciação): Calcula a potência de um número.

```python
2%2
```

### Operadores de Comparação:

*  `==` (igual a): Verifica se dois números são iguais.
* `!=` (diferente de): Verifica se dois números são diferentes.
* `<` (menor que): Verifica se um número é menor que o outro.
* `>` (maior que): Verifica se um número é maior que o outro.
* `<=` (menor ou igual a): Verifica se um número é menor ou igual ao outro.
* `>=` (maior ou igual a): Verifica se um número é maior ou igual ao outro.

## 4. Operadores lógicos

Os operadores lógicos são utilizados para avaliar expressões lógicas. Alguns exemplos são:
* `and` (E): retorna `True` se ambas as condições forem verdadeiras.
* `or` (OU): retorna `True` se pelo menos uma das condições for verdadeira.
* `not` (NÃO): inverte o valor da condição, ou seja, se a condição for verdadeira, `not` a tornará falsa, e vice-versa.
* `if` (SE) usado para executar um bloco de código somente se uma condição for verdadeira (avaliada como `True`). Ele permite que você tome decisões com base em condições.
* `else` (senão): frequentemente usado em conjunto (`if-else`) com `if`. Ele permite que você execute um bloco de código alternativo quando a condição do if for avaliada como `False`.

Além dos operadores lógicos, frequentemente é necessário realizar iterações entre as estruturas de dados. Para isso são utilizadas estruturas de fluxo, sendo a mais comum o loop `for` (Para, em português).

Antes de combiná-los com `if`, vamos ver `and`, `or` e `not` isoladamente, atuando sobre valores booleanos simples:

```python
choveu_hoje = True
solo_esta_seco = False

print(choveu_hoje and solo_esta_seco)  # E - as duas precisam ser True
print(choveu_hoje or solo_esta_seco)   # OU - basta uma ser True
print(not solo_esta_seco)              # NÃO - inverte o valor
```

Exemplos de uso de operadores lógicos:

Primeiro, o `if` sozinho: um bloco que só executa quando a condição é verdadeira.

```python
temperatura = 25  # suponha que aqui seja a temperatura atual do ar
limite_temperatura = 25 # suponha que quando a temperatura tiver maior que a temperatura limite, deve-se irrigar

if temperatura > limite_temperatura:
    print("É hora de irrigar")
```

Agora o `if/else`: dois caminhos possíveis, um para quando a condição é verdadeira e outro para quando é falsa.

```python
# Agora vamos considerar a umidade do solo como critério para ligar a irrigação
umidade_solo = 10  # Suponha que seja a umidade atual do solo
limite_umidade = 20

if umidade_solo < limite_umidade:
    print("É hora de irrigar")
else:
    print("A umidade do solo está adequada, não é necessário irrigar.")
```

Combinando duas condições com `and` — as duas precisam ser verdadeiras:

```python
# Agora considere que as plantas precisam estar dentro de um intervalo de temperatura e ter luz solar
temperatura = 28  # Suponha que seja a temperatura atual
luz_solar = True  # Suponha que há luz solar

limite_temperatura_min = 25
limite_temperatura_max = 30

if temperatura >= limite_temperatura_min and temperatura <= limite_temperatura_max and luz_solar:
    print("Condições ideais para o crescimento das plantações.")
```

Combinando condições com `or` — basta uma ser verdadeira:

```python
# Considere que as irrigações devem ser feitas considerando a umidade do solo e a previsão de chuva
previsao_chuva = False
umidade_solo = 18  # Suponha que seja a umidade atual do solo
limite_umidade = 20

if previsao_chuva or umidade_solo < limite_umidade:
    print("É recomendável irrigar devido à falta de umidade no solo ou à falta de previsão de chuva")
```

```python
# Suponha que a quantidade de chuva em milímetros seja dada por:
quantidade_chuva = 80

# Estabelecendo um limite mínimo de chuva necessário para o milho (em milímetros):
limite_minimo_chuva = 100

if quantidade_chuva >= limite_minimo_chuva:
    print("A quantidade de chuva é adequada para o plantio de milho.")
else:
    print("A quantidade de chuva é insuficiente para o plantio de milho. Considere irrigar")
```

Repetição de código com o loop `for`:

```python
# Exemplo de uso do loop for
for numero in range(1, 6):
    print(numero)
```

Quando existe uma condição intermediária, é necessário utilizar o `elif` (else if), além do `if` e do `else`. Isso permite controlar o fluxo de um programa com base em várias condições. Exemplo:

```python
# Suponha que você tenha a temperatura média atual em graus Celsius
temperatura = 28

# A faixa ideal para temperatura do ar para cultura do pimentão está entre 25 e 32 °C
# Para temperaturas abaixo de 25 °C, a condição é classificada como inadequada;
# Para temperaturas entre 25 e 32 °C, a condição é adequada;
# Para temperaturas acima de 32 °C, a condição é inadequada. Assim:

if temperatura < 25:
  print(" Temperatura inadequada")
elif temperatura >= 25 and temperatura <= 32:
  print("Temperatura adequada")
else:
  print("Temperatura inadequada")
```

### Recapitulando

* `and`, `or`, `not` combinam condições booleanas;
* `if` executa um bloco somente quando a condição é `True`;
* `if/else` cobre os dois caminhos possíveis;
* `elif` permite testar várias condições em sequência;
* `for` repete um bloco de código para cada elemento de uma lista (ou intervalo).

```python
# A temperatura ideal para esta cultura deve estar entre 20 e 30 ºC
# Assim, queremos saber quantos dias foram adequados e quantos foram inadequados
limite_minimo = 20
limite_maximo = 30

# Cria listas para armazenar os dias com temperaturas adequadas e inadequadas para o plantio de milho
dias_adequados = []
dias_inadequados = []
```

```python
# Iterando a lista de temperaturas e analisando
for temperatura in Tmed_obs:
    if temperatura >= limite_minimo and temperatura <= limite_maximo:  # Verifica se a temperatura é adequada
        dias_adequados.append(temperatura)  # Adiciona o dia à lista de dias adequados
    else:
        dias_inadequados.append(temperatura)  # Adiciona o dia à lista de dias inadequados
#dias_adequados
#dias_inadequados
```

### Recalculando a temperatura média com a biblioteca

Lá na seção 4, calculamos a temperatura média de forma manual como `(Tmax + Tmin) / 2`. A biblioteca já tem essa conta pronta na função `temp_media_extremos(Tmax, Tmin)`:

```python
Tmax = 32
Tmin = 18

# forma manual
media_manual = (Tmax + Tmin) / 2

# usando a biblioteca
media_agm = agm.temp_media_extremos(Tmax, Tmin)

print("Manual:", media_manual)
print("agrometeorologiapy:", media_agm)
```

### Funcionando com colunas inteiras de um DataFrame

O ponto forte de usar a biblioteca é que suas funções funcionam também com colunas de um `DataFrame` (do tipo `Series`, que vimos na seção 3) — o cálculo é feito para todas as linhas de uma vez, sem precisar de `for`:

```python
dados_estacao = {
    'Dia': [1, 2, 3, 4, 5],
    'Tmax': [31.0, 33.5, 29.0, 30.5, 32.0],
    'Tmin': [18.5, 19.0, 17.5, 18.0, 19.5]
}

df_estacao = pd.DataFrame(dados_estacao)
df_estacao['Tmed'] = agm.temp_media_extremos(df_estacao['Tmax'], df_estacao['Tmin'])
df_estacao
```
